```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2)
```

```python
kmeans.fit(data.drop(columns=['Private']))
```

```python
kmeans.cluster_centers_
```

```python
def converter(x):
    if x == 'Yes' :
        return 0
    else :
        return 1
```

```python
data['clusters'] = data['Private'].apply(converter)
data.head()
```

```python
from sklearn.metrics import classification_report

print(classification_report(
    data['clusters'],
    kmeans.labels_
))
```
